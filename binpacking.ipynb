# The Bin Packing Problem Examples

## Basic Example

- Suppose we have four items with sizes (or weights) 2, 3, 3, 2 in kgs.
- We have 3 bins with capaties 5,6,7 (kgs)
- Assign items to bins but use minimal number of bins
- In our example 2 bins are required to fit our items into these bins.
- A possible solution would be 1, 1, 2, 2. That means items 1 and 2 are put into bin 1
- and items 3 and 4 are put into bin 2.
- The last bin is not used.

\begin{aligned}
\min z = & y_1 + y_2 + y_3 \\
\text{Subject to:} & \\
& x_{11} + x_{12} + x_{13} = 1 \\
& x_{21} + x_{22} + x_{23} = 1 \\
& x_{31} + x_{32} + x_{33} = 1 \\
& x_{41} + x_{42} + x_{43} = 1 \\
& 2x_{11} + 3x_{21} + 3x_{31} + 2x_{41} \le 5y_1 \\
& 2x_{12} + 3x_{22} + 3x_{32} + 2x_{42} \le 6y_2 \\
& 2x_{13} + 3x_{23} + 3x_{33} + 2x_{43} \le 7y_3 \\
& x_{ij} \in \{0, 1\} \\
& y_j \in \{0, 1\} \\
& i = 1, 2, \ldots, 4 \\
& j = 1, 2, 3 \\
\end{aligned}

```julia
using Pkg; Pkg.activate(".");
```

```julia
using JuMP, HiGHS
```

```julia
begin
    caps = [5, 6, 7]
    model = Model(HiGHS.Optimizer)
    MOI.set(model, MOI.Silent(), true)
    @variable(model, y[1:3], Bin)
    @variable(model, x[1:4, 1:3], Bin)
    @objective(model, Min, sum(y))
    for i in 1:4
        @constraint(model, sum(x[i,:]) == 1)
    end 
    for j in 1:3
        @constraint(model, 2*x[1,j] + 3*x[2,j] + 3*x[3, j] + 2*x[4,j] <= caps[j] * y[j])
    end 
    optimize!(model)
end 
```
