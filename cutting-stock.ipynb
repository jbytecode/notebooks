# The Cutting Stock Problem

- Cables are 10 meters long.
- Order1: 3 meter x 20
- Order2: 4 meter x 15
- The objective is to minimize the unused parts of cuts

Cutting Patterns:

- 3-3-3  (total 9 meters - 1 meter unused)
- 3-3-4  (total 10 meters - no unused cables)
- 4-4    (total 8  - 2 meters unused)

- $x_1$: How many cuts will be done in pattern1
- $x_2$: How many cuts will be done in pattern2
- $x_3$: How many cuts will be done in pattern3

$$
\begin{aligned}
\min z = & x_1 + x_2 + x_3 \\
\text{Subject to:} & \\
& 3x_1 + 2x_2 + 0x_3 \ge 20 \;\; \text{Order1 requirements} \\
& 0x_1 + 1x_2 + 2x_3 \ge 15 \;\; \text{Order2 requirements} \\
& x_1, x_2, x_3 \in \{0, Z^+\} \\
\end{aligned}
$$

```julia
using Pkg; Pkg.activate("."); using JuMP, HiGHS
```

```julia
begin
    model = Model(HiGHS.Optimizer)
    MOI.set(model, MOI.Silent(), true)
    @variable(model, x[1:3] >= 0, Int)
    @objective(model, Min, sum(x))
    @constraint(model, 3x[1] + 2x[2] + 0*x[3] >= 20)
    @constraint(model, 0*x[1] + 1x[2] + 2x[3] >= 15)
    optimize!(model)
    println(value.(x))
end 
```

- $x_1 = 0$
- $x_2 = 10$
- $x_3 = 3$

- pattern1 is repeated 0 times
- pattern2 is repeated 10 times
- pattern3 is repeated 3 times
- if pattern2 is repeated 10 times, (3-3-4) ->  3 is repeated 20 times, 4 is repeated 10 times
- if pattern3 is repeated 3 time (4-4) -> 4 is repeated 6 times
- So count of 3 is 20
- so count of 4 is 10 + 6 = 16
- which satisfy the conditions
