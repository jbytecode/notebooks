# Dual Simplex Example

\begin{aligned}
\min z = & 100x_1 + 120x_2 \\
\text{Subject to:} & \\
& 3x_1 + 2x_2 \ge 36 \\
& 2x_1 + 4x_2 \ge 24 \\
& x_1, x_2 \ge 0 \\
\end{aligned}

- Making basic variables infeasible

\begin{aligned}
\min z = & 100x_1 + 120x_2 \\
\text{Subject to:} & \\
& -3x_1 - 2x_2 + s_1 = -36 \\
& -2x_1 - 4x_2 + s_2 = -24 \\
& x_1, x_2, s_1, s_2 \ge 0 \\
\end{aligned}

- Initial simplex table:

|        |    x1    |      x2    |      s1    |     s2     | Solution     |
| :---:  | :------: | :-------:  | :--------: |  :-------: | :---------:  |
|   z    |    -100  |   -120     |       0    |    0       |    0         |
|  s1    |   -3     |    -2      |       1    |    0       |    -36       |
|  s2    |   -2     |    -4      |       0    |    1       |    -24       |

- $s_1 = -36$ which is infeasible
- $s_2 = -24$ which is infeasible
- Since |-36| > |-24|, $s_1$ violates more.
- So $s_1$ exits the solution

- $x_1 \rightarrow \frac{-100}{-3}$: (33.3)
- $x_2 \rightarrow \frac{-120}{-2}$: (60)
- $s_1 \rightarrow \frac{0}{1}$: The denominator should be negative. Out of comparison.
- $s_2 \rightarrow \frac{0}{0}$: The denominator should be negative. Out of comparison.
- 33.3 is less than 60 and the denominators are negative, $x_1$ enters the solution

- In short $s_1$ exits, $x_1$ enters

|        |    x1    |      x2    |      s1    |     s2     | Solution     |
| :---:  | :------: | :-------:  | :--------: |  :-------: | :---------:  |
|   z    |    0     |  -160/3    |   -100/3   |    0       |  1200        |
|  x1    |    1     |   2/3      |     -1/3   |    0       |    12        |
|  s2    |    0     |  -8/3      |     -2/3   |    1       |    0         |

- $x_1 = 12$ which is feasible
- $s_2 = 0$ which is feasible
- No exiting variable.
- The final solution is the optimal solution.
- Solution = $\{x_1 = 12, x_2 = 0, z = 1200\}$

# Dual Simplex Example 2

\begin{aligned}
\min z = & 3x_1 + 2x_2 \\
\text{Subject to:} & \\
& 3x_1 + x_2 \ge 3 \\
& 4x_1 + 3x_2 \ge 6 \\
& x_1 + x_2 \le 3 \\
& x_1, x_2 \ge 0 \\
\end{aligned}

### Standard Form

\begin{aligned}
\min z = & 3x_1 + 2x_2 \\
\text{Subject to:} & \\
& -3x_1 - x_2 + s_1 = -3 \\
& -4x_1 - 3x_2 + s_2 = -6 \\
& x_1 + x_2 + s_3 = 3 \\
& x_1, x_2 \ge 0 \\
\end{aligned}

### Initial table 

|       |    x1   |    x2   |    s1     |    s2     |    s3     |   Solution   |
| :---: | :---:   |  :---:  |   :---:   |  :---:    |  :---:    |   :---:      |
|   z   |   -3    |    -2   |     0     |    0      |    0      |    0         |
|  s1   |   -3    |    -1   |     1     |    0      |    0      |    -3        |
|  s2   |   -4    |    -3   |     0     |    1      |    0      |    -6        |
|  s3   |    1    |     1   |     0     |    0      |    1      |     3        |

- s1 violates non-negativity
- s2 violates non-negativity
- s3 is fiasible
- So s1 and s2 are candidates
- s2 violates more! So s2 is exiting the solution...

- $x_1$: $\frac{-3}{-4}$: 0.75
- $x_2$: $\frac{-2}{-3}$: 0.66
- $s_1$: $\frac{0}{0}$: indeterminate
- $s_2$: $\frac{0}{1}$: The denominator should be negative
- $s_3$: $\frac{0}{0}$: indeterminate
- $x_1$ and $x_2$ are candidates. The smallest value is 0.66 so $x_2$ enters.

|       |    x1   |    x2   |    s1     |    s2     |    s3     |   Solution   |
| :---: | :---:   |  :---:  |   :---:   |  :---:    |  :---:    |   :---:      |
|   z   |   -1/3  |     0   |     0     |  -2/3     |    0      |    4         |
|  s1   |   -5/3  |     0   |     1     |  -1/3     |    0      |    -1        |
|  x2   |   4/3   |     1   |     0     |  -1/3     |    0      |    2         |
|  s3   |  -1/3   |     0   |     0     |   1/3     |    1      |     1        |

- $s_1$ violates the non-negativity condition so it's infeasible.
- It's only the one candidate to exit.
- So $s_1$ is exiting.

- $x_1$: $\frac{-1}{3} / \frac{-5}{3}$: 0.1999
- $x_2$: $0/0$: indeterminate
- $s_1$: $0/1$: The denominator should be negative
- $s_2$: $\frac{-2}{3} / \frac{-1}{3}$: 2
- $s_3$: $0/0$: indeterminate 
- so $x_1$ enters

|       |    x1   |    x2   |    s1     |    s2     |    s3     |   Solution   |
| :---: | :---:   |  :---:  |   :---:   |  :---:    |  :---:    |   :---:      |
|   z   |   0     |     0   |   -1/5    |  -3/5     |    0      |    21/5      |
|  x1   |   1     |     0   |   -3/5    |   1/5     |    0      |    3/5       |
|  x2   |   0     |     1   |    4/5    |  -3/5     |    0      |    6/5       |
|  s3   |   0     |     0   |   -1/5    |   2/5     |    1      |    6/5       |

- No violation of non-negativity conditions
- This is the final table.
- Solution $=\{x_1=\frac{3}{5}, x_2=\frac{6}{5}, z = \frac{21}{5}\}$
