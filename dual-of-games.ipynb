## Linear program for the row player

\begin{aligned}
\max z = & g \\
\text{Subject to:} & \\
& 2p_1 + 4p_2 \ge g \\
& 3p_1 + 5p_2 \ge g \\
& p_1 + p_2 = 1 \\
& p_1, p_2 \ge 0 \\
& g \in \mathcal{R} \\
\end{aligned}

- Since $g$ is unbounded, let $g = a - b$ where $a, b \ge 0$ 

\begin{aligned}
\max z = & a-b \\
\text{Subject to:} & \\
& 2p_1 + 4p_2 \ge a-b \\
& 3p_1 + 5p_2 \ge a-b \\
& p_1 + p_2 = 1 \\
& a, b, p_1, p_2 \ge 0 \\
\end{aligned}

\begin{aligned}
\max z = & a-b \\
\text{Subject to:} & \\
& 2p_1 + 4p_2 -a + b - s_1 = 0 \\
& 3p_1 + 5p_2 -a + b - s_2 = 0 \\
& p_1 + p_2 = 1 \\
& a, b, p_1, p_2, s_1, s_2 \ge 0 \\
\end{aligned}

\begin{aligned}
\min w = & 0y_1 + 0y_2 + y_3 \\
\text{Subject to:} & \\
& 2y_1 + 3y_2 + y_3 \ge 0\\
& 4y_1 + 5y_2 + y_3 \ge 0\\
& -y_1 - y_2 \ge 1\\
& y_1 + y_2 \ge -1\\
& -y_1 \ge 0 \\
& -y_2 \ge 0 \\
& y_1, y_2, y_3 \in \mathcal{R} \\
\end{aligned}

\begin{aligned}
\min w = & y_3 \\
\text{Subject to:} & \\
& -2y_1 - 3y_2  \le y_3\\
& -4y_1 - 5y_2  \le y_3\\
& y_1 + y_2 = -1\\
& y_1 \le 0 \\
& y_2 \le 0 \\
& y_1, y_2, y_3 \in \mathcal{R} \\
\end{aligned}

- let $x_1 = -y_1$ and $x_2 = -y_2$ 

\begin{aligned}
\min w = & y_3 \\
\text{Subject to:} & \\
& 2x_1 + 3x_2  \le y_3\\
& 4y_1 + 5y_2  \le y_3\\
& x_1 + x_2 \le -1\\
& y_1 \le 0 \\
& y_2 \le 0 \\
& y_1, y_2, y_3 \in \mathcal{R} \\
\end{aligned}
